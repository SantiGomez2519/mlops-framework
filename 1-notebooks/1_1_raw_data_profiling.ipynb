# 1.1 Raw Data Profiling

Load and inspect the raw house data to understand its structure, types, and quality before any transformations.

**Steps:**
1. Load raw CSV
2. Inspect data structure and types
3. Check for missing values
4. Explore unique values and categorical distributions

```python
import pandas as pd
import numpy as np
import warnings

warnings.filterwarnings('ignore')
```

## 1.1.1 Load Raw Data

```python
RAW_DATA_PATH = '../shared/data/raw/house_data.csv'

df = pd.read_csv(RAW_DATA_PATH)
print(f'Loaded {df.shape[0]} rows and {df.shape[1]} columns')
df.head(10)
```

## 1.1.2 Inspect Data Structure

```python
print('=== Data Types ===')
print(df.dtypes)
print(f'\n=== Memory Usage ===')
print(f'{df.memory_usage(deep=True).sum() / 1024:.2f} KB')
```

```python
df.describe().round(2)
```

## 1.1.3 Check for Missing Values

```python
print('=== Missing Values ===')
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Count': missing, 'Percentage': missing_pct})
print(missing_df[missing_df['Count'] > 0] if missing.any() else 'No missing values found.')
```

## 1.1.4 Unique Values and Categorical Distributions

```python
print('=== Unique Values per Column ===')
for col in df.columns:
    print(f'{col}: {df[col].nunique()} unique values')
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
print('=== Categorical Column Values ===')
for col in categorical_cols:
    print(f'\n{col}:')
    print(df[col].value_counts())
```
